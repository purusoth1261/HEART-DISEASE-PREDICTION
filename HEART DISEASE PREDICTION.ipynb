Divide the data into train and test
Step 4: Split dataset into training set and test set

```python
from sklearn.model_selection import train_test_split
```

Every data set has input as well as output variables. In this case study, Weight and Height are our input variables. And “Gender” is our output variable. So let's separate our data frame df into X and y. X will store the input variable. y will store the output variable.

Drop the Target or output variable. we have to identify the gender of the variable based on height and weight so drop the gender variable

```python
X=df.drop('heart disease',axis=1)
y=df['heart disease']
```

```python
X
```

```python
y
```

Now, to proceed further we have to divide our data into the training part and testing part. Training part will be used to train our model. The testing part will be used to test our model whether it gives correct predictions or not. and also to measure the model's performance. So divide X and y into X_train,X_test and y_train,y_test respectively.

```python
# Now call train_test_split function and pass X,y to divide them 
# into 4 parts as X_train,X_test and y_train,y_test
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.3, random_state=0)
# test size is 0.2 that is 20% so train size will be 80% automatically.
# random state is 112 or any numeric number. 
# Every random state number has a different fashion to select data rows randomly.
# random_state simply sets seed to the random generator, 
# so that your train-test splits are always deterministic.
```

Now we have to enter into 3rd phase of Machine learning life cycle. In this phase, we apply an algorithm to our dataset. So that our model gets trained and able to observe and analyse patterns within data. So lets apply DecisionTreeClassifier algorithm in this case study.

Step 5: Fit Classification Model

```python
# LOGISTIC REGRESSION
```

```python
from sklearn.linear_model import LogisticRegression
Logit_Model=LogisticRegression() #solver='lbfgs', max_iter=200)
Logit_Model.fit(X_train,y_train)
ans=Logit_Model.score(X_train,y_train)
print("The logistic Regression Training score",ans*100)
```

```python

#EVALUATION METRICS
from sklearn.metrics import accuracy_score 
prediction=Logit_Model.predict(X_test)
score = accuracy_score(y_test,prediction)
print("The logistic Regression Testing score",score*100)
```

```python
disease_pred = Logit_Model.predict([[67,1,160,286]])
print("My Predicted Value is is ",disease_pred)
```
